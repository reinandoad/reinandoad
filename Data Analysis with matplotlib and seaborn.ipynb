### **Nomor 1**

Tampilkan visualisasi yang menggambarkan persebaran harga rumah secara keseluruhan!

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.read_csv('melb_data.csv')
```

```python
df = pd.read_csv('melb_data.csv')
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['Price'], bins=50, kde=True)
plt.title('Distribusi Harga Rumah', fontsize=15)
plt.xlabel('Price')
plt.ylabel('Frekuensi')
plt.show()
```

Insight dari histogram tersebut:
1. Distribusi skewed right (miring ke kanan), artinya:
- Sebagian besar rumah berada pada rentang harga rendah hingga menengah.
- Ada sedikit rumah dengan harga sangat tinggi (outlier), sehingga distribusi miring ke kanan.

2. Mayoritas harga rumah
- Puncak histogram (frekuensi tertinggi) ada di kisaran harga di bawah 1 juta AUD. Artinya, pasar rumah di Melbourne didominasi oleh rumah-rumah dengan harga menengah ke bawah.

3. Outliers
- Histogram menunjukkan beberapa rumah dengan harga jauh lebih tinggi dibanding mayoritas.

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['Price'])

plt.title('Boxplot Harga Rumah', fontsize=15)
plt.xlabel('Price')
plt.show()
```

Insight:
1. Outlier banyak dan ekstrem
- Terlihat banyak titik di luar whisker kanan → ini rumah-rumah dengan harga sangat mahal

2. Median tidak di tengah box
- Median agak ke kiri, maka data harga positively skewed (dikonfirmasi oleh histogram).

### **Nomor 2**

Tampilkan visualisasi yang menggambarkan rata-rata harga rumah berdasarkan Region Name!

Jenis visualisasi yg digunakan: Barplot
Karena ingin membandingkan antar kategori (Regionname) terhadap nilai numerik (Price), maka barplot sangat cocok digunakan.

```python
plt.figure(figsize=(12,10))

# Barplot Mean
plt.subplot(2, 1, 1)
sns.barplot(x='Regionname', y='Price', data=df, estimator='mean', ci=None)
plt.title('Rata-rata Harga Rumah per Region', fontsize=15)
plt.xlabel('')
plt.ylabel('Rata-rata Harga')
plt.xticks(rotation=45)

# Barplot Median
plt.subplot(2, 1, 2)
sns.barplot(x='Regionname', y='Price', data=df, estimator='median', ci=None)
plt.title('Median Harga Rumah per Region', fontsize=15)
plt.xlabel('Regionname')
plt.ylabel('Median Harga')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Insight:
1. Wilayah dengan harga rumah tertinggi adalah Southern Metropolitan
- Baik mean maupun median harga tertinggi ditemukan di Southern Metropolitan.
- Artinya, kawasan ini memang tergolong elit secara konsisten, bukan hanya karena ada outlier mahal.

2. Wilayah dengan harga rumah terendah adalah Western Victoria
- Baik mean maupun median menunjukkan harga rumah paling rendah ada di Western Victoria.

3. Terdapat perbedaan besar antara mean dan median di beberapa region
- Misalnya di Eastern Metropolitan, mean lebih tinggi dibanding median -> mengindikasikan adanya outlier rumah sangat mahal.
- Harga rata-rata bisa menyesatkan karena terdongkrak oleh beberapa properti mahal.

### **Nomor 3**

Tampilkan visualisasi yang menggambarkan jumlah rumah untuk tiap tipe rumah (Type)!

Jenis Visualisasi yang Cocok: Barplot (Frekuensi Kategori)
Karena kita ingin mengetahui jumlah (count) dari masing-masing tipe rumah, maka barplot dengan sumbu Y sebagai jumlah dan sumbu X sebagai Type.

```python
type_counts = df['Type'].value_counts().reset_index()
type_counts.columns = ['Type', 'Jumlah']
```

```python
plt.figure(figsize=(8,5))
ax = sns.barplot(x='Type', y='Jumlah', data=type_counts)

for i in ax.containers:
    ax.bar_label(i, fmt='%d', label_type='edge', padding=3)

plt.title('Jumlah Rumah per Tipe Rumah', fontsize=15)
plt.xlabel('Tipe Rumah')
plt.ylabel('Jumlah')
plt.show()
```

Insight:
1. Tipe rumah terbanyak adalah 'h' (house), menunjukkan pasar dominan rumah tapak.
2. Tipe 'u' (unit/apartment) atau 't' (townhouse) lebih sedikit -> insight untuk kebutuhan pengembangan apartemen di masa depan.

### **Nomor 4**

Tampilkan visualisasi yang menggambarkan perubahan jumlah rumah yang dijual dari Jan 2016 - Des
2017!

Jenis Visualisasi yang Cocok: Lineplot (Time Series)
Karena ingin melihat perubahan jumlah penjualan rumah dari waktu ke waktu.

```python
df.info()
```

```python
# mengubah kolom date menjadi datetime
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df['Date']
```

```python
# mengambil tahun 2016 dan 2017 saja
mask = (df['Date'] >= '2016-01-01') & (df['Date'] <= '2017-12-31')
df_filtered = df.loc[mask]
df_filtered.head()
```

```python
# Buat kolom YearMonth (misal: 2016-01)
df_filtered['YearMonth'] = df_filtered['Date'].dt.to_period('M').astype(str)

# Hitung jumlah rumah yang dijual per bulan
monthly_counts = df_filtered['YearMonth'].value_counts().sort_index()
```

```python
# mengatur ukuran dari grafik
plt.figure(figsize=(12,6))

# membuat line plot
sns.lineplot(x=monthly_counts.index,
             y=monthly_counts.values,
             marker='o')

# membuat judul
plt.title('Penjualan Rumah tahun 2016-2017')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Rumah Terjual')
plt.xticks(rotation = 45)

plt.show
```

Insight:
1. Adanya fluktuasi penjualan dari tahun 2016 hingga tahun 2017
2. Adanya tren kenaikan di 2017. Meskipun fluktuatif, penjualan rumah di tahun 2017 secara umum lebih tinggi daripada 2016
2. Terlihat lonjakan besar di bulan Mei 2016, November 2016, dan Juli 2017
4. Puncak penjualan tertinggi di Juli 2017
3. Penjualan Rumah Menurun di Awal Tahun
- Januari 2016 dan 2017 memiliki jumlah penjualan rumah yang sangat rendah
- Kemungkinan disebabkan oleh efek libur akhir tahun atau tahun baru, di mana aktivitas jual beli rumah melambat

### **Nomor 5**

Tampilkan visualisasi yang menggambarkan perbandingan dari persebaran harga rumah (Price)
berdasarkan tipe rumah (Type)! Rumah di harga berapa untuk sebuah rumah bertipe "t"
(townhouse) bisa dikatakan sebagai outlier?

```python
# Plot boxplot harga rumah berdasarkan tipe)
my_pal = {"h": "g", "u": "b", "t": "m"}
sns.boxplot (data = df,
             x = 'Type',
             y = 'Price',
             palette = my_pal)

plt.title('Persebaran Harga Rumah berdasarkan Tipe Rumah', fontsize=15)
plt.xlabel('Tipe Rumah')
plt.ylabel('Harga Rumah')
plt.show()
```

Insight:
1. Tipe h (house) memiliki harga tertinggi dan paling bervariasi
- Median harga rumah tipe h lebih tinggi dari u dan t.
- Rentang (box dan whisker) jauh lebih lebar, menandakan variasi harga yang sangat besar.
- Banyak outlier ekstrem (hingga >8 juta AUD), menunjukkan keberadaan rumah-rumah yang sangat mahal dari kebanyakan rumah lainnya.

2. Tipe u (unit/apartment) adalah yang paling murah
- Median harga terendah dari semua tipe.
- Rentang harga sempit dan konsisten.

3. Tipe t (townhouse) berada di antara h dan u
- Median harga townhouse lebih tinggi dari unit, tetapi lebih rendah dari house.
- Ada beberapa outlier, tetapi tidak sebanyak h.

4. Outlier harga tinggi pada semua tipe rumah
Terutama pada tipe h dan t, terlihat banyak titik di atas whisker -> menunjukkan rumah mahal yang sangat berbeda dari kebanyakan.



```python
# mencari harga rumah yg dikategorikan sebagai outlier pada rumah tipe t
# outlier > Q3 + 1.5 * IQR
# outlier < Q1 - 1.5 * IQR**

df_t = df[df['Type'] == 't']
```

```python
# Hitung IQR
Q1 = df_t['Price'].quantile(0.25)
Q3 = df_t['Price'].quantile(0.75)
IQR = Q3 - Q1

# batas outlier
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Harga rumah tipe 't' dianggap outlier jika lebih dari {upper_bound:,.0f} AUD dan kurang dari {lower_bound:,.0f}")

```
